```python
import pandas as pd
colaboradores = pd.read_excel("Catalogo_Colaboradores.xlsx")
df = pd.DataFrame(colaboradores)

def revisarnulos():
  nulos = df.isnull().sum()
  print(nulos[nulos > 0])
  return

def crearcolumnas():
  df["SalarioDiario"] = df["Sueldo Neto Total"]/160
  df["CostoInterno"] = df["Sueldo Neto IMSS"] + df["Sueldo Complementos"]
  return

def Limpieza():
  df.drop(df.columns[0], axis=1, inplace=True)
  df.drop(columns=["Notas"], inplace=True)
  df["Departamento"] = df["Departamento"].fillna("Sin asignar")
  df.drop_duplicates(inplace=True)
  return

def estadisticasreferencia():
  SalarioPromedio = df["SalarioDiario"].mean()
  SueldoPromedio = df["Sueldo Neto IMSS"].mean()
  CompensacionPromedio = df["Sueldo Complementos"].mean()
  CostoInternoPromedio = df["CostoInterno"].mean()
  print("Salario promedio:", SalarioPromedio)
  print("Sueldo IMSS promedio:", SueldoPromedio)
  print("Compensación promedio:", CompensacionPromedio)
  print("Costo interno promedio:", CostoInternoPromedio)
  return

def actualiza():
  df.to_excel("archivocambiado.xlsx")
  print("Filas y columnas finales:", df.shape)
  return

revisarnulos()
crearcolumnas()
Limpieza()
estadisticasreferencia()
actualiza()
```
